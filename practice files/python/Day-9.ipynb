# Understanding Exceptions

    Exception handling in Python allows you to handle errors gracefully and take corrective actions without stopping the execution of the program. 

    This lesson will cover the basics of exceptions, including how to use try, except, else, and finally blocks.

##### What Are Exceptions?

    Exceptions are events that disrupt the normal flow of a program. 
    
    They occur when an error is encountered during program execution. Common exceptions include:

    - ZeroDivisionError: Dividing by zero.

    - FileNotFoundError: File not found.

    - ValueError: Invalid value.
    
    - TypeError: Invalid type.

    - NameType: NameError

# Syntax
## Key Points:

    try: block is where you put the code that may raise an exception.

    except Exception as e: catches the specific error and stores it in e so you can print or log it.

    Avoid using a bare except: — it’s better to catch specific exceptions (e.g., NameError for undefined variables):

```python
try: # try block to catch exceptions

    statement(s) # code that may raise an exception

except ExceptionType: # except block to handle specific exception
    
    statement(s) # code to handle the exception
```

```python
# Example of an exception:
try:
    a = c # This will raise a NameError because 'c' is not defined 

    print(a) # This will not be printed because the exception is raised before this line is executed.


except TypeError: # This will not catch the NameError exception because it is looking for a TypeError exception
    
    print("Variable c is not defined") # This will not be printed because the exception is not caught


except NameError: # This will catch the NameError exception and execute the code in the except block

    print("Variable 'c' is not defined. Please define 'c' before using it.") # This will be printed instead of the error message


print("This will be printed because the exception is handled and the program continues to execute.")
```

```python
# Example of an exception:

try:
    a = c # NameError: name 'c' is not defined

except NameError as ex: # handling the NameError exception and storing the exception message in variable ex = name 'c' is not defined

    print(ex) # printing the exception message stored in variable ex
```

```python
# Handling the exception using try-except block

try:  # try block to catch the exception
    print(my_list[0])
    print(my_list[1])
    print(my_list[2])
    print(my_list[3])
    print(my_list[4])
    print(my_list[5])
    print(my_list[6])  # This will raise an IndexError because there is no index 6 in the list

except IndexError as ie:  # except block to handle the IndexError exception
    print("Index out of range. Please provide a valid index between 0 and 5.")  # printing an error message
    print(ie)  # printing the exception message stored in variable ie

    
print("Program continues to execute...")  # this will be printed even after the exception is handled\
print("This will be printed because the exception is handled and the program continues to execute.")
```

```python
result = 1/0 # This will raise a ZeroDivisionError because you cannot divide by zero

print(result) # This will not be printed because the exception is raised before this line is executed
```

```python
# Handling the exception using try-except block

try:
    result = 1/0 # This will raise a ZeroDivisionError because you cannot divide by zero

    print(result) # This will not be printed because the exception is raised before this line is

except ZeroDivisionError as ex: # This will catch the ZeroDivisionError exception and execute the code in the except block

    print(ZeroDivisionError) # This will print the error message instead of the program crashing

    print(ex) # This will print the error message instead of the program crashing

    print("Please enter the denominator greater than 0") # This will be printed instead of the error message

print("This will be printed because the exception is handled and the program continues to execute.")


```

## Explanation

    try block

    Python executes the code inside the try block first.

    result = 1/0 tries to divide 1 by 0, which is not allowed and raises a ZeroDivisionError.

    except ZeroDivisionError as ex block

    Since a ZeroDivisionError is raised, Python jumps to this except block.

    The error object is stored in the variable ex.

    print(ex) displays the actual error message:

### Key Point

    try = code that might cause an error.

    except = code that runs if the error occurs.

    as ex gives you the actual error message for debugging or logging.

# try–except–else


```python
# Let's say we want to take input from the user and divide 1 by the input value. We can handle the exception if the user enters 0 or a non-numeric value.
# We can also use the else block to execute code that should run only if no exception occurs.
# The else block is optional and can be used to execute code that should run only if no exception occurs in the try block.
# The else block will not be executed if an exception is raised in the try block, even if the exception is handled in the except block.
# The else block is useful for code that should only run if the try block succeeds without any exceptions.
# Example:

try:
    num = int(input("Enter a number:"))  # This will raise a ValueError if the user enters a non-integer value

    result = 10 / num   # This will raise a ZeroDivisionError if the user enters 0

except ValueError: # This will catch the ValueError exception and execute the code in the except block

    print("That's not a valid number!")  # This will be printed instead of the error message

except ZeroDivisionError: # This will catch the ZeroDivisionError exception and execute the code in the except block

    print("You can't divide by zero!")  # This will be printed instead of the error message

else: # This will be executed only if no exception occurs in the try block

    print(f"the result is {result}")   # This will be printed only if the try block succeeds without any exceptions

print("This will be printed because the exception is handled and the program continues to execute.")


```

# Block behavior:

    try → runs the risky code.

    except → catches and handles errors.

    else → runs only if no exception occurs in the try block.

    finally → runs every time — even if there’s an error, a return, or a break.

# Summary Table

|   🧱 **Block**   | 🕓 **When it Runs**                            | 📝 **Purpose**                                            | 💡 **Example Use Case**                        |
| :--------------: | :--------------------------------------------- | :-------------------------------------------------------- | :--------------------------------------------- |
|   🟦 **`try`**   | Always runs first                              | Holds the code that might raise an exception              | User input, file operations, network calls     |
|  🟧 **`except`** | Runs **only if** an exception happens in `try` | Catch and handle specific or general errors               | Handle `ValueError`, `ZeroDivisionError`, etc. |
|   🟩 **`else`**  | Runs **only if no** exception occurs in `try`  | Put code that should run when everything goes fine        | Print results, continue normal flow            |
| 🟨 **`finally`** | **Always runs**, no matter what                | Do cleanup work like closing files or releasing resources | Close DB connections, print final message      |
