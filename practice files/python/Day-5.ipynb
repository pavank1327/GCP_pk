## For Loop

## Syntax of for loop in Python:

    for variable in sequence:
        # code block to execute
        
    variable → A temporary name that takes each value from the sequence (e.g., i).

    sequence → Can be a list, tuple, string, or something like range().

    Code block → Indented code that runs on each loop iteration.

```python
# This is a simple loop that prints numbers from 1 to 5
print(1) # This line prints the number 1
print(2) # This line prints the number 2
print(3) # This line prints the number 3
print(4) # This line prints the number 4
print(5) # This line prints the number 5
```

```python
# This is a simple loop that prints numbers from 0 to 5 using a for loop
for num in range(6): # This line starts a for loop that iterates over the numbers from 0 to 5
    print(num) # This line prints the current number in the loop
```

```python
for j in range(3):
    # Start for
    print(j)
```

```python
for j in range(3):
    if j == 2 :
        break;    #using break to break the loop for a particular condition
    # Start for
    print(j)
```

```python
for j in range(3):
    if j == 2 :
        continue;    #using continue at that number avoinding that number
    # Start for
    print(j)
```

```python
for j in range(3):
    if j == 2 :
        pass;    #using pass itworks as it is as for loop
    # Start for
    print(j)
```

## Explanation:

    range(5) → creates numbers 0, 1, 2, 3, 4.

    for i in range(5): → assigns each number from the range to i one by one.

    print(i) → runs in each iteration, printing the current value of i.

## Syntax of for loop:

    for variable in sequence:
        # code block to execute

    variable → temporary name (e.g., i) that takes each value from the sequence.

    sequence → usually created with range(start, stop, step) for number loops.

    Indented code runs for each value in the sequence.

```python
for i in range(10, 0, -1): # This line starts a for loop that iterates over the numbers from 10 to 1 with a step of -1 (counting downwards)
    print(i) # This line prints the current number in the loop
```

```python
for i in range(10, 0, -2):  # This line starts a for loop that iterates over the numbers from 10 to 1 with a step of -2 (counting downwards by 2)
    print(i)
```

## Explanation:

    range(10, 1, -1)

    Start = 10 → loop begins at 10

    Stop = 1 → loop stops just before 1 (1 is not included)

    Step = -1 → decrements by 1 each time
    → Sequence generated: 10, 9, 8, 7, 6, 5, 4, 3, 2

    for i in range(10, 1, -1):

    First i = 10 → print 10

    Next i = 9 → print 9

    Continues until i = 2.

    When i would become 1, loop stops because stop value is exclusive.

```python
for i in range(10,1,-3):
    print(i)
```

## Syntax of for loop with strings:

    for variable in string:
        # code block to execute

    variable → takes each character from the string one by one.

    string → any string value (e.g., "Hello").

    The code block runs once for each character in the string.

```python
full_name = "Satish Mudde" # This line assigns the string "Satish Mudde" to the variable 'full_name'

print(full_name) # This line prints the string stored in the variable 'full_name'

for ch in full_name: # This line starts a for loop that iterates over each character in the string 'full_name'
    print(ch) # This line prints the current character in the loop
 
```

## Explanation:

    The string "Satish Mudde" contains characters:
    s, a, t, i, s, h,  , m, u, d, d, e (notice the space between h and M).

    for i in str:

    First i = 's' → print s

    Next i = 'a' → print a

    This continues character by character, including the space.

    The loop ends after the last character e.

## While Loop

    The while loop continues to execute as long as the condition is True.

## Syntax of while loop:

    while condition:
        # code block to execute

    condition → a logical expression checked before every iteration.

    The loop continues as long as the condition is True.

    The indented code block runs repeatedly until the condition becomes False.


```python
for i in range(5): # This line starts a for loop that iterates over the numbers from 0 to 4
    print(i) # This line prints the current number in the loop
```

```python
num = 0 # This line initializes the variable 'num' with the value 0, which will be used as a counter in the while loop

while num < 5: # This line starts a while loop that continues as long as the value of 'num' is less than 5 True
    print(num) # This line prints the current value of 'num' in the loop
    num = num + 1  # This line increments the value of 'num' by 1 in each iteration of the loop
```

## Explanation:

    count = 0 → variable initialization.

    while count < 5:

    Checks the condition → 0 < 5 ✅ True → enters the loop.

    print(count) → prints the current value of count.

    count = count + 1 → increments count by 1 after each iteration.

    Loop repeats until count becomes 5 → then count < 5 ❌ False → loop ends.

## Loop Control Statements

## break

    for/while loop:
        if condition:
            break
        # remaining code inside loop

    break → immediately exits the loop when the condition is met, even if the loop hasn’t finished all iterations.

    Control jumps out of the loop and continues with the code after the loop.


```python
for i in range(10):
    print(i)    
```

```python
for i in range(10):
    if i == 5:
        break
    print(i)
```

## explanation:

    range(10) → generates numbers 0 to 9.

    Loop starts:

    i = 0 → condition i == 5 ❌ → print 0

    i = 1 → ❌ → print 1

    i = 2 → ❌ → print 2

    i = 3 → ❌ → print 3

    i = 4 → ❌ → print 4

    i = 5 → ✅ condition met → break → loop stops immediately.

    Numbers 6, 7, 8, 9 are never executed because the loop is terminated at i = 5.

## continue

## Syntax of continue statement

    for/while loop:
        if condition:
            continue
        # remaining code inside loop


    continue → skips the current iteration and jumps to the next loop iteration.

    Code after continue inside the loop is not executed for that iteration.

```python
for i in range(10):
    print(i)
```

```python
for i in range(10):
    if i == 5: # T 
        continue
    print(i)
```

## Explanation:

    range(10) → gives numbers 0 to 9.

    For each i, check i % 2 == 0:

    If True → number is even → continue → skip printing.

    If False → number is odd → print(i).

    So the loop skips even numbers and prints odd numbers only.

    Iteration flow:

    i = 0 → even → continue → skip print

    i = 1 → odd → print 1

    i = 2 → even → skip

    i = 3 → print 3

    ... continues till i = 9.

## pass

    The pass statement is a null operation; it does nothing.

## Syntax of pass statement:

    for/while loop:
        if condition:
            pass
        # remaining code


    pass → a placeholder statement that does nothing.

    It’s used when you need a statement syntactically (like inside if or loops) but don’t want any action to happen there.

    The loop continues normally after encountering pass.

```python
for i in range(10):
    if i == 5:  # T
        pass
    print(i)
```

## Explanation:

    range(5) → generates 0, 1, 2, 3, 4.

    For each i:

    If i == 3 → execute pass → does nothing → goes to print(i) next.

    For all other values, the if is skipped, and print(i) runs normally.

    Since pass doesn’t interrupt the flow, every number gets printed, including 3.

    Iteration flow:

    i = 0 → no if match → print 0

    i = 1 → print 1

    i = 2 → print 2

    i = 3 → pass (do nothing) → print 3

    i = 4 → print 4

## Nested loopss

    a loop inside a loop

## Syntax of Nested for loops:

    for outer_variable in outer_sequence:
        for inner_variable in inner_sequence:
            # code block to execute


    The outer loop runs first.

    For each iteration of the outer loop, the inner loop runs completely.

    Once the inner loop finishes, control goes back to the next iteration of the outer loop.

```python
for i in range(3):
    for j in range(2):
        print(f"i:{i} and j:{j}")
```

## Explanation:

    range(3) → i takes values 0, 1, 2.

    For each i, j loops through range(2) → 0, 1.

    print(f"i:{i} and j:{j}") prints the pair of current i and j values.

    Iteration flow:

    i = 0 → inner loop runs

    j = 0 → print i:0 and j:0

    j = 1 → print i:0 and j:1

    i = 1 → inner loop runs

    j = 0 → print i:1 and j:0

    j = 1 → print i:1 and j:1

    i = 2 → inner loop runs

    j = 0 → print i:2 and j:0

    j = 1 → print i:2 and j:1

## Examples- Calculate the sum of first N natural numbers using a while and for loop

    while loop  

```python
n = 10
sum = 0
count = 1

while count <= n:
    sum = sum + count
    #print(sum) # sum val
    count = count + 1

print("Sum of first 10 natural number:", sum)
```

## Explanation:

    n = 10 → we want to find the sum of numbers from 1 to 10.

    sum = 0 → stores the running total.

    count = 1 → starting number for the loop.

    while count <= n:

    Loop runs while count is less than or equal to 10.

    Each iteration:

    sum = sum + count → adds the current number to the total.

    count = count + 1 → moves to the next number.

    When count becomes 11 → condition 11 <= 10 is False → loop stops.

    print(...) displays the total sum.

    Iteration flow:

    1st → sum=0+1=1

    2nd → sum=1+2=3

    3rd → sum=3+3=6
    ...

    10th → sum=45+10=55

## Example- Prime numbers between 1 and 100

```python
for num in range(1, 101):
    if num > 1:  # skip 0 and 1 as they are not prime
        for i in range(2, num):
            if num % i == 0:  # if divisible, it's not prime
                break
        else:
            print(num)  # executes if no break occurred → prime number
```

## Explanation:

    Outer loop → iterates num from 1 to 100.

    if num > 1 → skips 1 (not prime).

    Inner loop → checks all numbers from 2 to num-1:

    If any i divides num completely (num % i == 0) → number is not prime → break.

    If the loop finishes without breaking, else runs → number is prime → print it.

| 🧠 **Concept**      | 🎯 **Purpose**                                                 | ⏰ **When to Use**                                              | 🧪 **Example**                                      |
| ------------------- | -------------------------------------------------------------- | -------------------------------------------------------------- | --------------------------------------------------- |
| 🔁 **for loop**     | Repeat code for a **fixed number of times** or over a sequence | When the number of iterations is **known**                     | `for i in range(5): print(i)`                       |
| 🔄 **while loop**   | Repeat code **while a condition is True**                      | When the number of iterations is **unknown / condition-based** | `while x < 5: print(x); x += 1`                     |
| ⛔ **break**         | **Exit** the loop immediately                                  | When you want to **stop** the loop based on a condition        | `if i == 5: break`                                  |
| ⏭️ **continue**     | **Skip** the current iteration and continue                    | When you want to **skip specific cases** but continue looping  | `if i == 2: continue`                               |
| 📝 **pass**         | **Do nothing**, acts as a placeholder                          | When code is **not ready** but syntax requires something       | `for i in range(3): pass`                           |
| 🧩 **nested loops** | Loop **inside another loop**                                   | When working with **multi-level data** or combinations         | `for i in range(2): for j in range(3): print(i, j)` |


# Summary

    🔁 for → fixed loops

    🔄 while → condition-based loops

    ⛔ break → exit early

    ⏭️ continue → skip & go next

    📝 pass → placeholder

    🧩 nested loops → for multi-dimensional or layered tasks
