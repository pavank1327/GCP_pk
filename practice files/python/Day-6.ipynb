## List

    Definition: 
    
    A list is a collection of ordered, mutable (can change), and heterogeneous (can hold different data types) items.

    Allows duplicates: ✅

    Ordered: ✅ (maintains insertion order)

    Mutable: ✅ (can add/remove/modify elements)



```python
# !/usr/bin/env python3
# This code demonstrates the use of different data types in Python, including integers, strings, floats, and booleans. It also shows how to create a tuple that contains multiple data types.
# The code first assigns values to variables of different data types and prints them. Then, it creates a tuple that contains all the previously defined variables and prints the tuple.

a = 10  # This variable 'a' is assigned the integer value 10. Integers are whole numbers without a decimal point.
b = "apple"  # This variable 'b' is assigned the string value "apple". Strings are sequences of characters.
c = 3.14  # This variable 'c' is assigned the float value 3.14. Floats are numbers with a decimal point.
d = True  # This variable 'd' is assigned the boolean value True. Booleans represent truth values.

e = a, b, c, d  # This variable 'e' is assigned a tuple that contains the previously defined variables a, b, c, and d. Tuples are ordered collections of items that can be of different types and are immutable (cannot be changed after creation).

print(a)
print(b)
print(c)
print(d)
print(e)

```

```python
list_val = [10,"apple",3.14,True]  # This variable 'list_val'
# assigned a list that contains multiple data types: an integer, a string, a float, and a boolean. Lists are ordered collections of items that can be of different types.

print(list_val) # This line prints the contents of the list 'list_val' to the console.
```

```python
# Creating a list

# A list can contain elements of different data types
# Lists are ordered, mutable [changeable], and allow duplicate values.
# Lists are defined using square brackets [].
# Lists can be nested, meaning you can have a list within a list.
# Lists support various methods for manipulation, such as append(), remove(), and sort().
# Lists are mutable, meaning you can change their content after creation.
# Example of a list:


my_list = [11, "apple", 3.14, True, 11, "apple"] # A list with duplicate values


# Accessing list

print(my_list) 

# Accessing elements

# Lists are zero-indexed, meaning the first element is at index 0.

print(my_list[1])        # apple
  
# Iterating
for item in my_list:
   print(item)
```

## Tuple

    Definition: 
    
    A tuple is an ordered, immutable collection of items.

    Allows duplicates: ✅

    Ordered: ✅

    Mutable: ❌ (cannot modify after creation)

```python
# Creating a tuple
# A tuple is similar to a list but is immutable(unchangeable) meaning you cannot change its content after creation.
# Tuples are defined using parentheses ().
# Tuples can contain elements of different data types.
# Tuples are ordered and allow duplicate values.
# Example of a tuple:

my_list = [10, "apple", 3.14, True] # A list containing an integer, a string, a float, and a boolean

my_tuple = (10, "apple", 3.14, True,"apple") # A tuple containing an integer, a string, a float, and a boolean

# Accessing tuple
print(my_tuple) 

my_list = [10, "apple"]   # A list with an integer and a string

my_tuple = (10, "apple")  # A tuple with an integer and a string
# Accessing elements
# Tuples are also zero-indexed, meaning the first element is at index 0.
print(my_tuple[1])      # 10    


# Iteration
for item in my_tuple:
    print(item)

```

## Dictionary (dict)

    Definition: 
    
    A dict is a collection of key-value pairs.

    Keys must be unique; values can repeat.

    Ordered: ✅ (from Python 3.7+ keeps insertion order)

    Mutable: ✅

```python
name = "Alice" #    This variable 'name' is assigned the string value "Alice". Strings are sequences of characters enclosed in quotes.
age = 30    # This variable 'age' is assigned the integer value 30. Integers are whole numbers without a decimal point.
dob = "1990-01-01"  # This variable 'dob' is assigned the string value "1990-01-01". It represents a date of birth in the format YYYY-MM-DD.
```

```python
# Creating a dictionary
# A dictionary is a collection of key-value pairs.
# Dictionaries are defined using curly braces {}.
# Each key is unique, and values can be of any data type.
# Dictionaries are unordered, meaning they do not maintain the order of elements.
# Example of a dictionary:


my_dict = {
         "name": "John", 
         "age": 25, 
         "city": "London"
      }

# Accessing dictionary
print(my_dict)  # This line prints the entire dictionary 'my_dict' to the console.


# Accessing values
print(my_dict["name"])       # John
print(my_dict["age"])        # 25
print(my_dict["city"])       # London

# Looping
for key, value in my_dict.items():
   print(key, ":", value)
```

## Set

    Definition: 

    A set is a collection of unordered, mutable, unique items.

    No duplicates: ✅

    Unordered: ❌ maintains no order

    Mutable: ✅ (but elements must be immutable types)

```python
# Creating a set
# A set is an unordered collection of unique elements.
# Sets are defined using curly braces {} or the set() function.
# Sets do not allow duplicate values, and any duplicates will be automatically removed.
# Sets are mutable, meaning you can add or remove elements after creation.
# Example of a set:


my_set = {10,1, 2,4, 3, 3}   # duplicates auto-removed

# Accessing set
print(my_set)

# print(my_set[0])  # True

# Iteration
for item in my_set:
    print(item)
```

```python
# Creating a set
mylist2 = ["a",'s','z','d','f','g','h','j','l','a']

print(mylist2)

my_tuple2 = ("a",'s','z','d','f','g','h','j','l','a')

print(my_tuple2)

my_set1 = {"a",'s','z','d','f','g','h','j','l','a'}   # duplicates auto-removed

print(my_set1)
```

| Feature        | **List** 📝              | **Tuple** 📌           | **Dict** 📚                                  | **Set** ✨                             |
| -------------- | ------------------------ | ---------------------- | -------------------------------------------- | ------------------------------------- |
| **Ordered**    | ✅ Yes                    | ✅ Yes                  | ✅ Yes (Python 3.7+)                          | ❌ No                                  |
| **Mutable**    | ✅ Yes                    | ❌ No (Immutable)       | ✅ Yes                                        | ✅ Yes                                 |
| **Duplicates** | ✅ Allowed                | ✅ Allowed              | ❌ Keys → No duplicates<br>✅ Values → Allowed | ❌ Not allowed (unique elements)       |
| **Indexed**    | ✅ Indexed by position    | ✅ Indexed by position  | ✅ Indexed by keys                            | ❌ Not indexed                         |
| **Syntax**     | `[]`                     | `()`                   | `{key: value}`                               | `{}`                     |
| **Common Use** | Sequences / dynamic data | Fixed data (no change) | Key–value lookups (mappings)                 | Unique items / fast membership checks |


# Summary:

    List → Best for ordered, changeable collections where duplicates are fine.

    Tuple → Similar to list but immutable (can’t modify after creation).

    Dict → Key–value storage, keys must be unique. Great for lookups.

    Set → Unordered collection of unique items, ideal for removing duplicates or quick membership tests.
