# Lambda Functions in Python

    Lambda functions are small anonymous functions defined using the **lambda** keyword. 
    
    They can have any number of arguments but only one expression. 
    
    They are commonly used for short operations or as arguments to higher-order functions.



```python
# define a function that takes two numbers as arguments and returns their sum
# Example usage:

def addition(a,b):
    return a+b
```

```python
# calling the function with two numbers and printing the result
result = addition(2,5)
print(result)
```

```python
print(type(addition))
```

# Syntax

    lambda arguments: expression is the basic syntax to create a lambda function in Python.

    lambda → keyword to define an anonymous (nameless) function.

    arguments → input parameters like in normal functions.

    expression → a single line of logic that gets evaluated and returned automatically

    lambda arguments: expression

# The map() Function in Python

    The map() function applies a given function to all items in an input list (or any other iterable) and returns a map object (an iterator). 
    
    This is particularly useful for transforming data in a list comprehensively.

```python
def square(x):
    return x*x # logic to calculate the square of a number

# square(4)
```

```python
square(4)
```

```python
lambda_square = lambda x:x*x # logic to calculate the square of a number
```

```python
lambda_square(4)
```

```python
list_numbers = [1, 2, 3, 4, 5, 6, 7, 8]
```

```python
lambda_square(1, 2, 3, 4, 5, 6, 7, 8)
```

# Using a for loop

### list_numbers = [1, 2, 3, 4, 5, 6, 7, 8]

```python
list_numbers = [1, 2, 3, 4, 5, 6]
```

```python

result = [] # to store the squares of the numbers in the list
for n in list_numbers: # iterate through each number in the list
    result.append(n * n) # calculate the square of the number and append it to the result list
print(result) # print the list of squares of the numbers in the original list
```

# Using map function

```python
# using map function to calculate the squares of the numbers in the list
def mapsquare(x): # input is a number
    print("this is mapsquare ")
    return x*x # logic to calculate the square of a number
```

```python
mapsquare(3)
```

```python
number_tuple = (1,2,3,4,5,6) # tuple of numbers

list(map(mapsquare,number_tuple))  # input for map function is a function and an iterable (list, tuple, etc.)
```

# Explanation
    numbers is a list of integers.

    map(square, numbers) applies the function square to each element in the numbers list.

    list(...) converts the map object into a regular list so you can see the result.

# Using List Comprehension (shorter)

```python
numbers = [1, 2, 3, 4, 5, 6, 7, 8] # using list comprehension to calculate the squares of the numbers in the list

result = [n * n for n in numbers] # using list comprehension to calculate the squares of the numbers in the list

print(result) # print the list of squares of the numbers in the original list
```

# Explanation

    numbers = [...]

    You created a list of numbers from 1 to 8.

    [n * n for n in numbers] → List Comprehension

    This is a compact way to build a new list by looping through each n in numbers.

    For each n, it calculates n * n (square of n) and stores the result in a new list.

# Map multiple iterables

```python
numbers1 = [1, 2, 3]
numbers2 = [4, 5, 6]

added_numbers = list(map(lambda x, y: x + y, numbers1, numbers2))
print(added_numbers)
```

# Explanation

    numbers1 and numbers2

    map(lambda x, y: x + y, numbers1, numbers2)

    Here, map() is taking two iterables (numbers1 and numbers2).

    The lambda function has two parameters: x (from numbers1) and y (from numbers2).

    On each iteration, map takes corresponding elements from both lists and applies the lambda:

            1st iteration → x=1, y=4 → 1+4 = 5
            2nd iteration → x=2, y=5 → 2+5 = 7
            3rd iteration → x=3, y=6 → 3+6 = 9
    
    list(...)

    Converts the map object into a list so you can see the result.

    print(added_numbers)

    Prints the final list.
