# Introduction to Functions

    Definition:

    A function is a block of code that performs a specific task.
    
    Functions help in organizing code, reusing code, and improving readability.

    def → used to define a function in Python.

    function_name → name you give to the function (e.g., add_numbers).

    (parameters) → values the function expects (can be empty too).

    """Docstring""" → optional description of what the function does (used for documentation).

    Function body → code that runs when the function is called.

    return expression → sends back a value to where the function was called.

```python
# Function definition
# This is a sample function definition in Python. You can replace the parameters, docstring,
# and function body with your own code.
# The function takes in parameters, performs some operations, and returns a result.
# Example usage:

def greet():
    print(f"Hello Welcome To the paradise")

```

# Keywords Arguments

    **kwargs → this is a variable-length keyword arguments parameter.

    The ** means the function can accept any number of key=value pairs.

    All those keyword arguments are collected into a dictionary called kwargs.

    kwargs.items() → returns each key–value pair from the dictionary.

    The for loop goes through each key and value and prints them in key:value format.

```python
def print_details1(**kwargs):
    print(kwargs.items())
    print(type(kwargs.items()))
```

```python
print_details1(name="Satish", age=28, city="Hyderabad")
```

```python
def print_details(**kwargs):
    print(kwargs.items())
    print(type(kwargs.items()))
    for key, value in kwargs.items():
        print(f"{key} = {value}")
```

```python
print_details(name="Satish", age=28, city="Hyderabad")
```

    *args → collects any number of positional arguments into a tuple.

    **kwargs → collects any number of keyword arguments into a dictionary.

    First for loop → prints each positional argument one by one.

    Second for loop → prints each keyword argument in key:value format.

```python
def print_details(**satish):
    print(satish.items())
    print(type(satish.items()))
    for key, value in satish.items():
        print(f"{key} = {value}")
        
print_details(name="Satish", age=28, city="Hyderabad")
```

```python
def print_details(*args, **kwargs):
    for val in args:
        print(f" Positional arument :{val}")
    
    for key, value in kwargs.items():
        print(f"{key}:{value}")

print_details(10, 20, 30,50, name="Satish", role="Engineer", city="Hyderabad",a=50)

```

```python
def print_details(*args, **kwargs):
    for val in args:
        print(f" Positional arument :{val}")
    
    for key, value in kwargs.items():
        print(f"{key}:{value}")

print_details(10, 20, 30,50, 100,  name="Satish", role="Engineer", city="Hyderabad",a=50, b = 100, c = 200)
```

| 🌟 **Feature**             | 🧰 **`*args`** (Positional)                                       | 🧭 **`**kwargs`** (Keyword)                                     |
| -------------------------- | ----------------------------------------------------------------- | --------------------------------------------------------------- |
| ✨ **Meaning**              | Variable-length **positional** arguments                          | Variable-length **keyword** arguments                           |
| 🪄 **Prefix**              | Single `*`                                                        | Double `**`                                                     |
| 🧱 **Data Type Collected** | Stored as a **tuple** 🟡                                          | Stored as a **dictionary** 🟢                                   |
| 📌 **Use Case**            | When you don’t know how many **positional values** will be passed | When you don’t know how many **key=value pairs** will be passed |
| 🧠 **Example Definition**  | `def func(*args):`                                                | `def func(**kwargs):`                                           |
| 🧪 **Example Call**        | `func(1, 2, 3)`                                                   | `func(name="Satish", age=28)`                                   |
| 📤 **Example Output**      | `(1, 2, 3)` 📝                                                    | `{'name': 'Satish', 'age': 28}` 📝                              |


# Summary:

    Use *args when you want to accept any number of positional values.

    Use **kwargs when you want to accept any number of key–value pairs.

    Use both if you want maximum flexibility in your function.
