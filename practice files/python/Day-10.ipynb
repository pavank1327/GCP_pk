# Iterators

    Iterators are advanced Python concepts that allow for efficient looping and memory management. 

    Iterators provide a way to access elements of a collection sequentially without exposing the underlying structure.


```python
# !/usr/bin/env python3
# This is a simple Python script that creates a list of numbers from 1 to 6 and prints it to the console.
# The script uses the built-in `print` function to display the contents of the list.
# The list is defined using square brackets and contains the integers 1 through 6, separated by commas.
# When you run this script, it will output the list to the console.
# To run this script, save it to a file (e.g., `my_list.py`) and execute it using Python 3 in your terminal or command prompt.
# Example usage:

my_list = [1,2,3,4,5,6] # This creates a list of numbers from 1 to 6 and assigns it to the variable `my_list`.

print(my_list) # This prints the contents of `my_list` to the console, which will output: [1, 2, 3, 4, 5, 6]
```

```python
# Accessing individual elements of the list using their index
# In Python, list indices start at 0, so the first element is accessed with index 0, the second with index 1, and so on.
# The following lines of code demonstrate how to access and print each element of the list individually using their respective indices.
# When you run this part of the script, it will print each element of the list on a new line, followed by "END" to indicate the end of the output.
# Note: If you try to access an index that is out of range (e.g., my_list[6]), it will raise an IndexError since there are only 6 elements in the list (indices 0 to 5).
# Example usage:

print(my_list[0])   # 1 
print(my_list[1])   # 2
print(my_list[2])   # 3
print(my_list[3])   # 4
print(my_list[4])   # 5
print(my_list[5])   # 6
```

```python
# Iterating through the list using a for loop
# A common way to access each element of a list is by using a for loop. The
# following code demonstrates how to iterate through the list `my_list` and print each element on a new line.
# The for loop automatically goes through each element in the list, assigning it to the variable `i` in each iteration, which is then printed to the console.
# When you run this part of the script, it will output each element of the list on a separate line.
# Example usage:


my_list=[1,2,3,4,5,6]  # This creates a list of numbers from 1 to 6 and assigns it to the variable `my_list`.

for i in my_list: # This starts a for loop that iterates through each element in `my_list`. In each iteration, the current element is assigned to the variable `i`.
    
    print(i) # This prints the current element (stored in `i`) to the console. Each element of the list will be printed on a new line.
```

```python
print(type(my_list)) # This prints the type of the variable `my_list` to the console, which will output: <class 'list'>, indicating that `my_list` is a list data structure in Python.
```

```python

my_list1 = [1,2,3] 
 
print((my_list1)) # This line prints the contents of `my_list1` to the console, which will output: [1, 2, 3], showing that `my_list1` is a list containing the integers 1, 2, and 3.

print(type(my_list1))  # This line checks the data type of `my_list1` using the `type` function, which will return `<class 'list'>`, confirming that `my_list1` is a list object in Python.

print(my_list1[0])   # This line accesses the first element of the list `my_list1` using index 0 and prints it to the console. Since `my_list1` contains [1, 2, 3], this will output: 1.

# print(my_list1[3])   # This line accesses the second element of the list `my_list1` using index 1 and prints it to the console. Since `my_list1` contains [1, 2, 3], this will output: 2.

# iterator

iterator_list_val = iter(my_list1) # This line creates an iterator object from the list `my_list1` using the built-in `iter()` function and assigns it to the variable `iterator_list_val`.

print(iterator_list_val) # This line prints the `iterator_list_val` variable to the console. Since `iterator_list_val` is an iterator object, it will output something like `<list_iterator object at 0x...>`, indicating that it is an iterator for the list `my_list1`.

print(type(iterator_list_val)) # This line checks the data type of `iterator_list_val` using the `type` function, which will return `<class 'list_iterator'>`, confirming that `iterator_list_val` is an iterator object for the list `my_list1`.


# print(iterator_list_val[0]) # This line attempts to access the first element of the iterator `iterator_list_val` using index 0. However, this will raise a TypeError because iterators do not support indexing. Iterators can only be accessed sequentially using the `__next__()` method or by using a loop.

print(iterator_list_val.__next__()) # This line calls the `__next__()` method on the iterator `iterator_list_val`, which returns the next element in the iteration. Since `my_list1` contains [1, 2, 3], this will output: 1, which is the first element of the list.

print(iterator_list_val.__next__()) # This line calls the `__next__()` method on the iterator `iterator_list_val`, which returns the next element in the iteration. Since `my_list1` contains [1, 2, 3], this will output: 2, which is the second element of the list.

print(iterator_list_val.__next__()) # This line calls the `__next__()` method on the iterator `iterator_list_val`, which returns the next element in the iteration. Since `my_list1` contains [1, 2, 3], this will output: 3, which is the third element of the list.

# print(iterator_list_val.__next__()) # This line calls the `__next__()` method on the iterator `iterator_list_val` again, but since there are no more elements left in the list to iterate over, this will raise a StopIteration exception, indicating that the iteration has been exhausted.
```

```python
type(iterator_list_val) # This line checks the data type of `iterator_list_val` using the `type` function, which will return `<class 'list_iterator'>`, confirming that `iterator_list_val` is an iterator object for a list.
```

```python

my_list3 = (3,4,5,6)

iterator3 = iter(my_list3) # This line creates an iterator from the list `my_list3` using the built-in `iter` function and assigns it to the variable `iterator3`.

#print(iterator3.__next__()) # This line calls the `__next__()` method on the iterator `iterator3`, which returns the next element in the list, which is 3.

#print(iterator3.__next__()) # This line calls the `__next__()` method on the iterator `iterator3` again, which returns the next element in the list, which is 4.

print(next(iterator3)) # This line calls the built-in `next()` function with the iterator `iterator3` as an argument, which is equivalent to calling `iterator3.__next__()`.
print(next(iterator3)) # This line calls the `next()` function again with the iterator `iterator3`, which will return the next element in the list, which is 5.
print(next(iterator3)) # This line calls the `next()` function again with the iterator `iterator3`, which will return the next element in the list, which is 6.
print(next(iterator3)) # This line calls the `next()` function again with the iterator `iterator3`, but since there are no more elements left in the list, it will raise a `StopIteration` exception, indicating that the iterator has been exhausted.
```

```python
type(iterator3) # This line checks the data type of `iterator3` using the `type` function, which will return `<class 'tuple_iterator'>`, confirming that `iterator3` is an iterator object for a tuple.
```

```python
# String iterator
my_string = "satish"

string_iterator = iter(my_string)

print(type(string_iterator)) # This line checks the data type of `string_iterator` using the `type` function, which will return `<class 'str_iterator'>`, confirming that `string_iterator` is an iterator object for a string.

print(next(string_iterator))  # Output: s

print(string_iterator.__next__()) # Output: a

print(next(string_iterator)) # Output: t

print(next(string_iterator)) # Output: i

print(next(string_iterator)) # Output: s

print(next(string_iterator)) # This line calls the `next()` function again with the `string_iterator`, which will return the next character in the string, which is 'h'.

# print(next(string_iterator))
```

# Generators
    Generators are a simpler way to create iterators. 
    
    They use the yield keyword to produce a series of values lazily, which means they generate values on the fly and do not store them in memory.

```python
# normal function
def ten():

    return 10   # return 10 value when the function is called return keyword 
```

```python
ten() # calling ten() function, which will return the value 10 when executed.
```

```python
type(ten()) 
```

```python
val_ten = ten() # function call to `ten()` and assigns the returned value (10) to the variable `val_ten`. After this line is executed, `val_ten` will hold the value 10.
```

```python
type(val_ten)
```

```python
print(val_ten)
```

```python
# generator function

def one():
    yield 1  # yield keyword 
```

```python
one() # calling one() function 
```

```python
type(one()) 
```

# Conclusion

    Iterators and generators are powerful tools in Python for creating and handling sequences of data efficiently. 

    Iterators provide a way to access elements sequentially, while generators allow you to generate items on the fly, making them particularly useful for handling large datasets and infinite sequences. Understanding these concepts will enable you to write more efficient and memory-conscious Python programs.

| 📝 **Feature**        | 🔁 **Iterator**                          | ⚡ **Generator**                                 |
| --------------------- | ---------------------------------------- | ----------------------------------------------- |
| ✍ **Definition**      | Uses `__iter__()` & `__next__()` methods | Uses `yield` in a function                      |
| 🧱 **Code Style**     | Class-based, more boilerplate 🧱         | Simple function, clean ✨                        |
| 🧠 **Memory**         | Can store entire data in memory 🧠       | Lazy evaluation → memory efficient 🌿           |
| 🚨 **Stop Condition** | Must raise `StopIteration` manually 🛑   | Auto `StopIteration` when function ends 🤖      |
| 🔄 **Reusability**    | Reusable after reset ♻️                  | One-time use, recreate after exhaust 🔸         |
| 🏃 **Performance**    | Slower for large data 🐢                 | Faster for large streams 🚀                     |
| 🧰 **Use Case**       | Complex stateful iteration 🧠            | Stream processing / ETL / Big Data pipelines 💧 |
