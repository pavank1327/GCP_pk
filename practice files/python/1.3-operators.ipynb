## Arithmethic Operation

```python
num1 = 10
num2 = 5

add_result = num1 + num2 #addiiton
print(add_result)

sub_result = num1 - num1  #substraction 
print(sub_result)

mult_result= num1 * num2 #multiplication
print(mult_result)

div_result= num1 / num2  #division
print(div_result)

floor_div_result = num1//num2 ## floor division
print(floor_div_result)

modulus_result = num1 % num2 #modulus operation
print(modulus_result)

exponent_result = num1 ** num2 ## Exponentiation
print(exponent_result)
```

```python
## Simple Calculator

# Simple calculator
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

# Performing arithmetic operations
addition = num1 + num2
subtraction = num1 - num2
multiplication = num1 * num2
division = num1 / num2
floor_division = num1 // num2
modulus = num1 % num2
exponentiation = num1 ** num2

# Displaying results
print("number1 = ",num1)
print("number2 = ",num2)
print("Addition:", addition)
print("Subtraction:", subtraction)
print("Multiplication:", multiplication)
print("Division:", division)
print("Floor Division:", floor_division)
print("Modulus:", modulus)
print("Exponentiation:", exponentiation)
```
