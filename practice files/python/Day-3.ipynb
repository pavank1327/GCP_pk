## Conditional Statements (if, elif, else, Nested if)


## If Statement



### Syntax 
    if condition:
        # code to execute if the condition is True


```python
age = 17


if age >= 18: # T/F

    print("✅ You are allowed to vote in the elections.") # ✅ You are allowed to vote in the elections. True

    # ✅ You are allowed to vote in the elections. 

```

## Explanation:

    The if statement checks a condition (age <= 18).

    If the condition is True, the code inside the if block runs.

    Here, since age is 20 (which is ≥ 18), the message is printed.

    If the condition is False, the block is skipped.

## if else

### Syntax 
    if condition:
        # code to execute if the condition is True
    else:
        # code to execute if the condition is False

```python
age = 17

if age >= 18: # T or F

    print("✅ You are allowed to vote in the elections.")  # True 

else:

    print("❌ You are a minor. Not eligible yet.")  # False
```

```python
age = 2

if age >=18: # T/F
    #  START TRUE BLOCK 
    print("✅ You are allowed to vote in the elections.")

    #  END TRUE BLOCK 
else:
    #  START FALSE BLOCK 
    print("❌ You are a minor. Not eligible yet.")
```

## Explanation:

    The if part runs when the condition is True.

    The else part runs when the condition is False.

    Since 16 is not ≥ 18, the else block executes.

## If–Elif–Else

### Syntax
    if condition1:
        # code to execute if condition1 is True
    elif condition2:
        # code to execute if condition2 is True
    elif condition3:
        # code to execute if condition3 is True
    else:
        # code to execute if all conditions are False

```python

age = 12

if age <= 13: # T/F

    print("👶 You are a child.")

elif age <= 18: # T/F

    print("🧒 You are a teenager.")

elif age <= 50: # T/F

    print("🧒 You are a adult.")

elif age <= 60: # T/F

    print("🧒 You are a adult.")

else:
    print("🧑 You are an Old.")

```

```python
age = 60

if age <= 13: # T
    #  START TRUE BLOCK 
    print("👶 You are a child.") # STOP
    #  END TRUE BLOCK 
elif age <=18: # T
    #  START TRUE BLOCK 
    print("🧒 You are a teenager.") # STOP
    #  END TRUE BLOCK
elif age <= 50: # T
    #  START TRUE BLOCK 
    print("🧒 You are a adult.") # STOP
    #  END TRUE BLOCK
else: # F
    #  START FALSE BLOCK 
    print("🧑 You are an Old.")
```

## Explanation:

    if checks the first condition.

    If it's False, elif (else if) checks the next condition.

    If both are False, else runs as a fallback.

    Here, 17 is not < 13, but it is < 18 → so the elif block executes.

## Quick Summary Table

| 🧠 **Concept**            | 🧾 **Keyword(s)**      | ⏰ **When to Use**                                                        |
| ------------------------- | ---------------------- | ------------------------------------------------------------------------ |
| ✅ **Single condition**    | `if`                   | Run code only when a specific condition is **True**                      |
| 🔄 **Two conditions**     | `if` + `else`          | Handle both **True** and **False** cases clearly                         |
| 🌿 **Multiple branches**  | `if` + `elif` + `else` | Check **multiple conditions** one by one in order                        |
| 🧭 **Step-by-step check** | **Nested** `if`        | When one condition **depends on another** (hierarchical decision making) |


## Nested Condiitonal Statements

        else:

    elif another_condition:
        # code for elif
    else:
        # code for else
        if nested_inner:
            # nested inside else


# You can place one or more if, elif, or else statements inside another if, elif, or else statement to create nested conditional statements.

    number even ,odd,negative

```python
num=int(input("Enter the number"))

if num  > 0: # F OUTTER IF
    print("The number is positive")

    if num%2 == 0: # F INNER IF
        print("The number is even")
    else:
        print("The number is odd")
else:
    print("The number is zero or negative")
```

## Explanation

    First if checks whether the number is positive (num > 0).

    If true, it prints the message and goes inside the nested if to check whether it's even or odd using num % 2 == 0.

    elif checks for the zero case separately, because zero is neither positive nor negative.

    else runs if the number is negative.

    Inside this block, there’s another nested if to check whether the negative number is divisible by 5 (num % 5 == 0) to show deeper nesting.

## Practical Examples

    Determine if a year is a leap year using nested condition statement


```python

year=int(input("Enter the year"))

if year%4==0:  # T/F
    if year%100 == 0: # T/F
        if year%400==0: # T/F
            print(year,"is a leap year")
        else:
            print(year,"is not a this leap year")
    else:
        print(year,"is a this leap year")

else:
    print(year,"is not a leap year")
```

```python
age = 20

if age < 0:
    print("num pos")
else:
    print("num pos")
    
```

## Explanation

    First if checks whether the year is divisible by 4 (year % 4 == 0).
    If this is false, it immediately prints that the year is not a leap year.

    If the year is divisible by 4, the program moves inside the nested if to check whether the year is divisible by 100 (year % 100 == 0).
    This is because century years need an extra check.

    Second nested if (inside the 100 check) verifies whether the year is divisible by 400 (year % 400 == 0).

    If true → the year is a leap year.

    If false → the year is not a leap year (e.g., 1900).

    Else inside the 100 check runs if the year is not divisible by 100, which means it's a normal leap year (e.g., 2024).

    Final else (outermost) runs if the year is not divisible by 4, meaning it’s definitely not a leap year.

## Enhanced Version (Using Logic Instead of Nesting)

```python
year = int(input("Enter a year: "))

if year > 0:
    if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
        print(f"✅ {year} is a leap year.")
    else:
        print(f"❌ {year} is NOT a leap year.")
else:
    print("⚠️ Please enter a valid positive year.")
```

# Summary:

    Use simple if for one condition.

    Use if-else for either/or situations.

    Use if-elif-else for multiple possible outcomes.

    Use nested if when conditions are hierarchical or dependent.
