# Decorators

    Decorators are a powerful and flexible feature in Python that allows you to modify the behavior of a function or class method. 

    They are commonly used to add functionality to functions or methods without modifying their actual code. 

    This lesson covers the basics of decorators, including how to create and use them.

## Explanation

# Closures functions

```python
# welcome function with parameter

def main_welcome(msg): # function definition
   
    def sub_welcome_method(): # nested function definition

        print("Welcome to the advance python course") # nested function body

        print(msg) # nested function body

        print("Please learn these concepts properly") # nested function body

    return sub_welcome_method() # sub_welcome_method is returned to the caller when main_welcome is called
# return "test" # main_welcome returns the value to the caller when main_welcome is called
```

```python
main_welcome("Welcome everyone") # main_welcome is called and the value returned by main_welcome is printed to the caller
```

## Explanation

    ✅ def main_welcome(msg): → Defines a function named main_welcome that takes one parameter msg.

    msg will be used inside the nested function to display custom content.

    : marks the beginning of the function block.

    ✅ def sub_welcome_method(): → This defines an inner (nested) function inside main_welcome.

    This function has no parameters.

    It is defined but not executed until explicitly called.

    This is a classic example of a nested function in Python.

    ✅ print("Welcome to the advance python course") → Prints a welcome message when the inner function is called.

    ✅ print(msg) → Uses the parameter from the outer function inside the inner function.

    This shows how the inner function can access variables from the outer function’s scope (closure behavior).

    ✅ print("Please learn these concepts properly") → Prints an additional instruction.

    ✅ return sub_welcome_method() →

    Here, we are calling the inner function sub_welcome_method() and returning its result.

    Since sub_welcome_method() has only print statements and no return, it implicitly returns None.

    So the outer function executes the inner function and then returns None.

```python
def main_welcome(func):
   
    def sub_welcome_method():
        print("Welcome to the advance python course")
        func("Welcome everyone to this tutorial")
        print("Please learn these concepts properly")
    return sub_welcome_method()
```

```python
main_welcome(print) # main_welcome is called and the value returned by main_welcome is printed to the caller
```

## Explanation

    ✅ def main_welcome(func): → Defines a function main_welcome that takes a function as a parameter named func.

        This allows us to pass any function to main_welcome and then use it inside.

        This is an example of higher-order functions in Python.

    ✅ def sub_welcome_method(): → Defines a nested function inside main_welcome.

        No parameters are needed here.

        This function will use func (passed from outside) to print a message.

    ✅ print("Welcome to the advance python course") → Prints the initial welcome message when sub_welcome_method runs.

    ✅ func("Welcome everyone to this tutorial") →

        Calls the function that was passed into main_welcome.

        In this example, print is passed, so this line becomes:

        This is how we dynamically call a function that was passed as an argument.

    ✅ print("Please learn these concepts properly") → Prints the final instruction after calling the passed function.

    ✅ return sub_welcome_method() →

        This calls the nested function sub_welcome_method immediately and returns its result.

        Since the inner function only prints and doesn’t return anything, the return value is None, but the prints are executed when main_welcome is called.

    ✅ main_welcome(print) →

        Here, the built-in print function is passed as the func argument.

        So inside sub_welcome_method, when func(...) is called, it executes print(...).

        This shows how you can pass functions as arguments and execute them inside other functions.

# Decorator

```python
def main_welcome(func):
   
    def sub_welcome_method():
        print("Welcome to the advance python course")
        func()
        print("Please learn these concepts properly")
    return sub_welcome_method()
```

```python
def coure_introduction():
    print("This is an advanced python course...............")

coure_introduction()
```

```python
main_welcome(coure_introduction)
```

## Explanation

    ✅ def main_welcome(func): →

        Defines a function named main_welcome that takes another function as a parameter func.

        This allows us to pass a function to main_welcome, which can later be called inside.

        This is a typical pattern used in function decorators and higher-order functions.
    
    ✅ def sub_welcome_method(): →

        Defines a nested function inside main_welcome.

        This function has no parameters.

        It contains three steps: print a message, call the passed-in function, and print another message.

    ✅ print("Welcome to the advance python course") →

        Prints the initial welcome message when the inner function runs.

    ✅ func() →

        Calls the function that was passed into main_welcome.

        Since func is expected to be a function reference, this line actually executes that function.

    ✅ print("Please learn these concepts properly") →

        Prints the closing message after the passed function has executed.
    
    ✅ return sub_welcome_method() →

        Calls the nested function sub_welcome_method immediately and returns its result.

        Since the inner function only prints and doesn’t return anything, the overall return value is None, but all print statements are executed at that moment.

    ✅ def coure_introduction(): →

        Defines a regular function that prints:

    ✅ coure_introduction() →

        Calls the coure_introduction function directly.

        It simply executes the print statement inside.

```python
main_welcome(coure_introduction)
```

## Explanation

    ✅ main_welcome(coure_introduction) →

        Here we are calling the main_welcome function and passing the function coure_introduction as an argument.

        Notice: We are passing coure_introduction without parentheses → this means we are passing the function itself, not its return value.

        Inside main_welcome, this gets assigned to the parameter func.
    
    Inside main_welcome

        When main_welcome is called, the following happens:

        1️⃣ sub_welcome_method is defined inside main_welcome.
        2️⃣ Then return sub_welcome_method() is executed → this calls the nested function immediately.

```python
@main_welcome
def coure_introduction():
    print("This is an advanced BIG DATA course!!!!!!!!!!!!!!!!!!!")
```

```python
coure_introduction()
```

## Explanation

    ✅ @main_welcome → This is a decorator.

        In Python, the @decorator_name syntax is shorthand for:

        It means the function coure_introduction is passed as an argument to main_welcome right after it’s defined.

        The return value of main_welcome(coure_introduction) replaces the original coure_introduction function.

        main_welcome receives the original coure_introduction as its func argument.

        Inside main_welcome, it defines sub_welcome_method, then calls it immediately because of return sub_welcome_method().

        So the welcome message, original function, and closing message all execute right away.

```python
def my_decorator_satish(func):
    def wrapper():
        print("Something is happening before the function is called.")
        func()
        print("Something is happening after the function is called.")
    return wrapper
```

## Explanation

    ✅ def my_decorator(func): →

        This defines a decorator function named my_decorator.

        It takes another function func as an argument.

        The idea is to wrap extra behavior around func without changing its original code.

    ✅ def wrapper(): →

        This is the inner function that actually adds extra behavior.

        It will execute code before and after calling the original function func().

        It’s called wrapper because it wraps the original function’s behavior.

    ✅ print("Something is happening before the function is called.") →

        This runs before the original function.

        Useful for logging, setup, validation, etc.

    ✅ func() →

        Executes the original function that was passed to the decorator.

        This is the core action that remains unchanged.

    ✅ print("Something is happening after the function is called.") →

        This runs after the original function.

        Useful for cleanup, logging, or any post-processing.

    ✅ return wrapper →

        The decorator returns the inner function object, not the result of calling it.

        This is crucial — it allows the decorator to replace the original function with the wrapped version, while still letting you call it later.

```python
@my_decorator_satish # decorator
def main_hello(): # function
    print("Hello!")
```
